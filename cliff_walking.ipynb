<details>
  <summary markdown='span'>
  💡 Tuple unpacking?
  </summary>

  If a function returns a tuple, you can immediately save the different elements of the tuple in different variables.

  Example:

  Imagine you have a function:

  ```python
  def surface_and_circumference(a, b):
    """Returns surface and circumference of a rectangle with
    length `a` and width `b`."""
    return a*b, 2*a + 2*b
  ```

  Instead of this:

  ```python
  result = surface_and_circumference(4, 2)
  surface = result[0]
  circumference = result[1]
  ```

  You can immediately do this:

  ```python
  surface, circumference = surface_and_circumference(4, 2)
  ```

  If you're only going to use the surface in the rest of your code, it's common practice to use `_` for the other return values. It's a sign for other programmers that you're throwing away the remaining values.

  Example:
  
  ```python
  surface, _ = surface_and_circumference(4, 2)
  # Here goes code that only needs surface
  ```

</details>

````python

# Returns a tuple:
new_state, reward, terminated, truncated, info = env.step(action)

print('Nuevo estado:', new_state)
print('Recompensa obtenida:', reward)
print('Terminado (done):', terminated)
print('Truncado:', truncated)
print('Información adicional:', info)
print(env.render())
````

**📍 New State**: This number represents the state of the agent after taking the specified action. In the `CliffWalking` environment, the state corresponds to a specific position in the environment grid that the agent has moved to.

**💸 Reward**: The reward value indicates the immediate feedback given by the environment as a result of the agent's action. In many grid-based environments, a negative reward like this typically signifies a penalty, suggesting that the action taken may not be optimal or is penalized to encourage other strategies.

**🚦 Done**: The boolean value indicates whether the episode has ended. In this case, `False` means the episode is still ongoing, and the agent hasn't reached a terminal state (like a goal or a pitfall) that would end the episode.

### Section 4: Training a RL algorithm


First, you will prepare the `CliffWalking-v0` environment for training with Stable Baselines3 models. This includes proper setup and wrapping to ensure compatibility with RL algorithms.

#### 🧱 1. Load the Environment:
- Use `gym.make()` to create the `CliffWalking-v0` environment.  
- Set:
  - `is_slippery=True` → Adds randomness to movements (more realistic).  
  - `render_mode='human'` → Enables visual feedback during interaction.  


````python
env = gym.make('CliffWalking-v0', render_mode = 'ansi', is_slippery = False)
````

#### ❄️ Understanding `is_slippery=True`

Setting `is_slippery=True` in `CliffWalking-v0` adds randomness to the agent’s actions — introducing a chance that the agent slips and moves in an unintended direction. Training with `is_slippery=True` builds agents that can handle uncertainty — a critical skill for real-world applications.

#### 🎯 Why Add Stochasticity?

- **🌍 Realism** → Simulates uncertainty found in real-world environments.  
- **🛡️ Robustness** → Helps agents learn more reliable, adaptable strategies.  
- **🔥 Challenge** → Makes the task harder and more interesting to solve.

#### 🧵 2. Wrap the Environment

Wrapping your environment with [`DummyVecEnv`](https://stable-baselines3.readthedocs.io/en/v1.0/guide/vec_envs.html#dummyvecenv) is a small but critical step — it ensures your RL setup runs smoothly with Stable Baselines3.

Go ahead and wrap the environment using [`DummyVecEnv`](https://stable-baselines3.readthedocs.io/en/v1.0/guide/vec_envs.html#dummyvecenv) from Stable Baselines3.  👇

````python
# Add this import at the start of the notebook:
from stable_baselines3.common.vec_env import DummyVecEnv

# Wrap the environment to be compatible with SB3
env = DummyVecEnv([lambda: env])  # Lambda function to ensure correct environment handling
````

#### 🧵 More on `DummyVecEnv`

`DummyVecEnv` is a wrapper from Stable Baselines3 that makes Gym environments compatible with RL models by vectorizing them.

#### ⚙️ What Does It Do?

- **📦 Standardizes the API**  
  Ensures the environment matches the expected format for Stable Baselines3 algorithms.

- **📊 Enables Batch Processing**  
  Even with one environment, interactions are treated like a batch — essential for scaling later with tools like `SubprocVecEnv`.

- **🔗 Ensures Compatibility**  
  Wraps `reset()` and `step()` so they work correctly inside training loops.

#### 3. Initialize a DQN Model from SB3

Now that your environment is ready, you need to initialize the model.  In this case, you will configure a Deep Q-Network (DQN) using Stable Baselines3.  DQN uses a neural network to estimate Q-values — predicting the expected reward for each action in a given state.

#### 📝 Steps to Follow:
- Import `DQN` from `stable_baselines3`. Add it to the cell at the top of the notebook.
- Initialize an instance of the DQN model with the following parameters:
    - Use `'MlpPolicy'` — a basic neural network that maps observations to actions.  
    - Set the `env` to your environment and `verbose=1` for detailed training logs.
    - Add the `tensorboard_log` parameter to track training metrics. We'll use it later to follow the training with TensorBoard.

````python
model = DQN('MlpPolicy', env, verbose=1, tensorboard_log='dqn_tensorboard/')
````

### Section 5: Efficient Training Without Rendering

In this section, you will go through the entire process of preparing the environment, loading and training the model, but without visual rendering. Disabling rendering during the training phase can significantly speed up the training process since it reduces computational overhead.

#### 📝 Steps to Follow:

#### 🧱 1. Load and Prepare the Environment
- Use `render_mode=None` to disable visual output.

#### ⚙️ 2. Configure and Initialize the DQN Model
- Use `'MlpPolicy'` and `DummyVecEnv` as before.  
- Keep `verbose=1` if you want log output.
- We'll add a TensorBoard logging location `./dqn_tensorboard` so we can follow the training with TensorBoard.

#### ⏱️ 3. Set Training Parameters and Train
- Increase timesteps for better learning (e.g., `total_timesteps = 100_000`).  
- Call `.learn()` to start training.

#### 💾 4. Save the Trained Model
- Use `.save("dqn_cliffwalking_fast")` to store your model.

By following these steps, you efficiently train and save a DQN model without the added computational overhead of rendering the environment. This approach is particularly useful when training complex models or when using limited computational resources.

````python
# Load environment
env = gym.make('CliffWalking-v0', render_mode = None)

# Wrap the environment to be compatible with SB3
env = DummyVecEnv([lambda: env])

# Load the model, match it with the environment, set a policy, add a logging location
model = DQN('MlpPolicy', env, verbose=1, tensorboard_log="./dqn_tensorboard")

# Set training timesteps
total_timesteps = 100_000

# Train the model
model.learn(total_timesteps = total_timesteps)

# Define model save path
model_save_path = "dqn_cliffwalking_fast"

# Save the trained model
model.save(model_save_path)
````

#### 📊 Understanding Training Log Metrics

While training a reinforcement learning model, several metrics help track learning progress and performance.

#### 🎲 Rollout Metrics

- **exploration_rate** → Probability of taking a random action.  
  - High = more exploration  
  - Low = more exploitation  

#### ⏱️ Time-related Metrics

- **episodes** → Number of completed episodes.  
- **fps** → Frames per second (how fast training is running).  
- **time_elapsed** → Total time since training began (in seconds).  
- **total_timesteps** → Total number of steps taken in the environment.

#### 🧠 Training Metrics

- **learning_rate** → Size of updates to model weights.  
  - Lower = slower but more stable learning  
- **loss** → Prediction error of the model.  
  - Decreasing loss = learning is working  
- **n_updates** → Number of times the model has updated its weights.

#### 🔍 How to Interpret Them

- **⬇️ exploration_rate** → The agent is shifting from exploration to learned behavior.  
- **⚡ High fps** → Training is running efficiently.  
- **📉 Decreasing loss** → The model is improving and making fewer errors.
